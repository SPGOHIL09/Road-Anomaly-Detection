```python
import os
import cv2
import pandas as pd
import numpy as np
import torch
import ultralytics
import onnx


from ultralytics import YOLO
```

```python
if torch.backends.mps.is_available():
    device = "mps"
elif torch.cuda.is_available():
    device = "cuda"
else:
    device = "cpu"
print(f"Using device: {device}")
```

```python
model = YOLO("runs/detect/train3/weights/best.pt")

model.to(device)

test_image_path = "/Users/spgohil/Developer/Projects/DatasetNinja_RDD2022_MIX/test/img"

test_images = [os.path.join(test_image_path, img) for img in os.listdir(test_image_path) if img.endswith(".jpg")]

os.makedirs("predictions", exist_ok=True)

for img_path in test_images:
    results = model.predict(img_path,device=device, conf=0.3, iou=0.55, imgsz=640, augment=True)

    result_img = results[0].plot()
    save_path = os.path.join("predictions", os.path.basename(img_path))
    cv2.imwrite(save_path, result_img)
    print(f"Saved prediction to {save_path}")
    
    cv2.imshow("Prediction", result_img)
    cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
print("🎉 Predictions completed and saved in 'predictions' folder!")
```
